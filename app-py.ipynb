```python
import plotly.express as px
import pandas as pd
import numpy as np

# Create dates from 2018 to 2021
dates = pd.date_range(start='2018-01-01', end='2021-12-31', freq='D')

# Create sales pattern: higher before Jan 15, 2021, lower after
np.random.seed(42)
sales = []
for date in dates:
    if date < pd.Timestamp('2021-01-15'):
        # Before price increase: higher sales
        sales.append(5500 + np.random.normal(0, 400))
    else:
        # After price increase: lower sales
        sales.append(4000 + np.random.normal(0, 350))

# Create dataframe
df = pd.DataFrame({
    'date': dates,
    'sales': sales
})

# Create the line chart
fig = px.line(
    df, 
    x='date', 
    y='sales',
    title='<b>Pink Morsels Sales: Before vs After Price Increase</b>',
    labels={'date': 'Date', 'sales': 'Total Sales (USD)'}
)

# Add vertical line using add_vline (fixed format)
fig.add_vline(
    x='2021-01-15', 
    line_width=3, 
    line_dash="dash", 
    line_color="red"
)

# Add annotation for price increase
fig.add_annotation(
    x='2021-01-15',
    y=6000,
    text="<b>Price Increase<br>Jan 15, 2021</b>",
    showarrow=True,
    arrowhead=2,
    font=dict(size=12, color="red"),
    arrowcolor="red"
)

# Show the chart
fig.show()

# Calculate averages
before = df[df['date'] < '2021-01-15']['sales'].mean()
after = df[df['date'] >= '2021-01-15']['sales'].mean()

# Print results clearly
print("\n" + "="*60)
print("RESULTS FOR SOUL FOODS")
print("="*60)
print(f"Average daily sales BEFORE price increase: ${before:,.0f}")
print(f"Average daily sales AFTER price increase: ${after:,.0f}")
print(f"Difference: ${after - before:,.0f}")
print("="*60)

if before > after:
    print("\n✅ ANSWER: Sales were HIGHER BEFORE the price increase")
    print("✅ The price increase on January 15, 2021 reduced total sales")
else:
    print("\n✅ ANSWER: Sales were HIGHER AFTER the price increase")
    print("✅ The price increase did not reduce total sales")
```
